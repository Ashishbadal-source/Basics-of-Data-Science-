# Binomial distribution

```python
def fact(x):
    if x==0:
        return 1
    else:
        return x*fact(x-1)
```

```python
def combination(n,r):
    return fact(n)//(fact(r)*fact(n-r))
```

```python
n = 20
p=0.5
x = np.arange(0, n+1)
list=[]
for x in range(n+1):
    list.append(combination(n,x)*(p**x)*(1-p)**(n-x))
    
plt.vlines(np.arange(0, n+1), 0, list, colors='b', lw=5)
plt.plot(np.arange(0, n+1), list, 'bo', ms=8)
plt.title('Binomial Distribution n=20, p=0.5')
plt.xlabel('Number of successes')   
plt.ylabel('Probability')
plt.show()    
```

# Poisson distribution

```python
n = 30
p = 0.5
k_values = np.arange(0, n + 1)
lambda_val = n * p  

probabilities = []
for k in k_values:
    probabilities.append((np.exp(-lambda_val) * (lambda_val ** k)) / fact(k))

plt.vlines(k_values, 0, probabilities, colors='g', lw=5)
plt.plot(k_values, probabilities, 'go', ms=8)
plt.xlabel('k')
plt.ylabel('P(X = k)')
plt.title('Poisson Distribution (λ = n * p)')
plt.show()

```

# Normal distribution

```python
import math

n = 20
p = 0.5
mean_value = n * p
standard_deviation = math.sqrt(n * p * (1 - p))

x = np.linspace(0, n, 200)

pdf = (1 / (standard_deviation * math.sqrt(2 * math.pi))) * np.exp(-0.5 * ((x - mean_value) / standard_deviation) ** 2)

plt.plot(x, pdf, 'r-', lw=2)
plt.title('Normal Distribution Approximation')
plt.xlabel('x')
plt.ylabel('Probability Density Function')
plt.show()
```

Binomial: Discrete. Values only at integer counts.

Poisson: Discrete. Values only at integer counts.

Normal: Continuous. Density defined for all real x.

---> Discrete distributions produce isolated probabilities at integers. Their shape approximates a smooth curve only when n (or λ) is large.

---> The normal is continuous and smooth. For symmetric parameters (p=0.5) the binomial is symmetric and when variance is moderate/large it looks bell-shaped like the normal.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial as fact, comb as combination

n = 15
p = 0.15
λ = n * p
k = np.arange(0, n + 1)

poisson_pmf = [(np.exp(-λ) * (λ ** i)) / fact(i) for i in k]
binomial_pmf = [combination(n, i) * (p ** i) * ((1 - p) ** (n - i)) for i in k]

plt.figure(figsize=(10,6))
plt.bar(k - 0.18, binomial_pmf, width=0.35, color="#FF092E", alpha=0.7, edgecolor='black', label='Binomial (n=15, p=0.15)')
plt.bar(k + 0.18, poisson_pmf, width=0.35, color="#038022", alpha=0.6, edgecolor='black', label=f'Poisson (λ={λ:.2f})')

plt.plot(k, binomial_pmf, 'o-', color='navy', lw=2, ms=6)
plt.plot(k, poisson_pmf, 'o--', color='darkred', lw=2, ms=6)

plt.title(f'Poisson Approximation to Binomial Distribution\nn={n}, p={p}, λ={λ:.2f}', fontsize=14, pad=15)
plt.xlabel('k (Number of Successes)', fontsize=12)
plt.ylabel('Probability', fontsize=12)
plt.xticks(k)
plt.grid(True, linestyle='--', alpha=0.4)

plt.legend(loc='upper right', fontsize=11, frameon=True)
plt.tight_layout()
plt.show()
```

i) For n = 100 , p = 0.05, the two PMFs(Binomial and Poisson) almost completely overlap. 

ii) Both distributions peak near k=2 (the mean) and decline symmetrically on both sides.

iii) The difference between individual probabilities is very small — visually, the bars nearly coincide.

when n is large and p is small with λ=np constant, the Poisson distribution is a good approximation of the Binomial distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

n = 10      
p = 0.5      
lambda_val = 3  
sample_size = 1000  

import pandas as pd
data = pd.read_csv("iris.csv")

mean_normal = data['SepalLengthCm'].mean()
std_normal = data['SepalLengthCm'].std()

binom_samples = np.random.binomial(n, p, sample_size)
poisson_samples = np.random.poisson(lambda_val, sample_size)
normal_samples = np.random.normal(mean_normal, std_normal, sample_size)

def plot_stats(samples, title):
    plt.hist(samples, bins=30, color='green', edgecolor='black')
    plt.title(f'Histogram of {title} samples')
    plt.xlabel('Value')
    plt.ylabel('Frequency')
    plt.show()
    
    mean_val = np.mean(samples)
    var_val = np.var(samples)
    print(f"{title} Mean: {mean_val:.2f}")
    print(f"{title} Variance: {var_val:.2f}\n")

plot_stats(binom_samples, "Binomial")
plot_stats(poisson_samples, "Poisson")
plot_stats(normal_samples, "Normal")
```

```python
df = pd.read_csv("diabetes.csv")
data.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
```

```python
data = df.select_dtypes(include=['int64', 'float64'])
cols = data.columns
corr_matrix = pd.DataFrame(index=cols, columns=cols, dtype=float)

for i in cols:
    for j in cols:
        x = data[i].values
        y = data[j].values
        mean_x = sum(x) / len(x)
        mean_y = sum(y) / len(y)
        numerator = sum((x[k] - mean_x) * (y[k] - mean_y) for k in range(len(x)))
        denominator = math.sqrt(sum((x[k] - mean_x)*2 for k in range(len(x))) * sum((y[k] - mean_y)*2 for k in range(len(y))))
        corr = numerator / denominator if denominator != 0 else 0
        corr_matrix.loc[i, j] = corr

print("\nManual Correlation Matrix:")
print(corr_matrix.round(2))
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(corr_matrix.astype(float), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Manual Correlation Matrix - Diabetes Dataset")
plt.show()
```

```python
threshold = 0.9
corr_pairs = corr_matrix.unstack()
high_corr = corr_pairs[(abs(corr_pairs) > threshold) & (abs(corr_pairs) < 1)]
print("\nHighly Correlated Feature Pairs (>|0.9|):")
print(high_corr)

X = data.values
X_meaned = X - np.mean(X, axis=0)
cov_mat = np.cov(X_meaned, rowvar=False)
eigen_values, eigen_vectors = np.linalg.eig(cov_mat)
sorted_index = np.argsort(eigen_values)[::-1]
eigen_values = eigen_values[sorted_index]
eigen_vectors = eigen_vectors[:, sorted_index]
k = 2
eigenvector_subset = eigen_vectors[:, :k]
X_reduced = np.dot(X_meaned, eigenvector_subset)

print("\nOriginal shape:", X.shape)
print("Reduced shape:", X_reduced.shape)
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_reduced[:, 0], X_reduced[:, 1], c=df['Outcome'], cmap='coolwarm', edgecolor='k', alpha=0.7)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA Visualization (2 Components)")
plt.colorbar(label='Diabetes Outcome')
plt.show()
```

```python
explained_variance_ratio = eigen_values / sum(eigen_values)
print("\nExplained Variance Ratio (Top 2 PCs):")
print(explained_variance_ratio[:2])
```
