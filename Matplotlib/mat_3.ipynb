# 3-D Graphs

# 3-D Scatter plot

```python
import matplotlib.pyplot as plt
import pandas as pd
batter = pd.read_csv("batter.csv")
```

```python
fig = plt.figure(figsize=(10,7))
ax = plt.subplot(projection='3d')
ax.scatter3D(batter['runs'],batter['avg'],batter['strike_rate'],color="red")
ax.set_title("Average and strike rate")
ax.set_xlabel("Runs")
ax.set_ylabel("Average")
ax.set_zlabel("Strike Rate")
```

# 3-D Line plot

```python
x = [0,5,22,12]
y = [ 0,5,25,16]
z = [0,15,33,4]
fig = plt.figure()
ax = plt.subplot(projection="3d")
ax.scatter3D(x,y,z,s=[100,100,100,100])   # s is for size of the dot 
ax.plot3D(x,y,z,color="red")
```

# 3-D Surface plot

```python
import numpy as np
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx , yy = np.meshgrid(x,y)
xx.shape
yy.shape
zz = xx**2 + yy ** 2
zz
```

```python
fig = plt.figure(figsize=(8,8))
ax = plt.subplot(projection="3d")
p = ax.plot_surface(xx,yy,zz,cmap="viridis")
fig.colorbar(p)
```

```python
zz = np.sin(xx) + np.cos(yy)
```

```python
fig = plt.figure(figsize=(7,7))
ax = plt.subplot(projection="3d")
q = ax.plot_surface(xx,yy,zz,cmap="viridis")
plt.colorbar(q)
plt.show()
```

```python
z1 = np.sin(xx) + np.log(yy**2)
```

```python

fig = plt.figure(figsize=(9,6))
ax = plt.subplot(projection="3d")
r = ax.plot_surface(xx,yy,z1,cmap="viridis")
fig.colorbar(r)
plt.show()
```

# Countour plot => plot 3D graph into 2D

```python
import numpy as np
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
xx , yy = np.meshgrid(x,y)
zz = xx**2 + yy ** 2
fig = plt.figure(figsize=(10,7))
ax = plt.subplot()
p = ax.contour(xx,yy,zz,cmap="viridis")
fig.colorbar(p)                           # colors in the graph indicate the depth 
```

# Heat Map

```python
delivery = pd.read_csv("IPL_Ball_by_Ball_2008_2022.csv")
delivery.head()
```

```python
temp_df = delivery[(delivery["ballnumber"].isin([1,2,3,4,5,6]))& (delivery["batsman_run"]==6)]
```

```python
grid = temp_df.pivot_table(index="overs",columns="ballnumber",values="batsman_run",aggfunc ="count")
```

```python
plt.figure(figsize=(15,7))
plt.imshow(grid)
plt.yticks(delivery['overs'].unique(),list(range(1,21)))    # convert the overs from 1 to 20 rather than 0 to 19 which is index specific .
plt.xticks(np.arange(0,6),list(range(1,7)))                 # convert the balls from 1 to 6 rather than 0 to 5
plt.colorbar()
```
